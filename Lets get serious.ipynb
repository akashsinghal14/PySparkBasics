```
df = spark.read.format("csv")\
    .option("header","true")\
    .option("inferschema","true")\
    .option("mode","FAILFAST")\
    .load("/FileStore/tables/sales.csv")
```

```
df.head()
```

```
df.show()
```

```
df.createOrReplaceTempView("sales")
result = spark.sql("select * from sales limit 5")
result.show()
```

```
df.select(df.CustomerName).show()
```

```
# you can also use indexes to show columns
df.select(df.columns[:4]).show()
```

```
# add columns to existing df
adding_col = df.withColumn(
                'Total Cost', df.UnitPrice*df.Quantity
)
adding_col.show()
```

```
# filtering in pyspark

more_than_one_quantity = df.filter(df.Quantity >1)
more_than_one_quantity.show()
```

```
# products more than 5000
adding_col = df.withColumn(
                'Total Cost', df.UnitPrice*df.Quantity
)
high_value_products = adding_col.filter(adding_col["Total Cost"] > 3000)
high_value_products.show()
high_value_products.count()
```

```
# group by use
from pyspark.sql.functions import avg, col, desc, count, max

df.groupBy("OrderDate")\
    .agg(count("OrderDate").alias("Counts")).show()
# df.groupBy("OrderDate").count().show()
```

```
# Order by use
from pyspark.sql.functions import avg, col, desc, count, max

# Group by 'OrderDate' and calculate the maximum 'UnitPrice'
df.groupBy("OrderDate")\
    .agg(count("UnitPrice").alias("max_unit_price"))\
    .orderBy("max_unit_price", ascending=False)\
    .show()
```

```
# drop duplicates

drop_d = df.dropDuplicates(['OrderDate'])
drop_d.show()
drop_d.count()
```
